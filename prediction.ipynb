```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
## load the trained model, scaler pickle file and the one hot encoded 
model=load_model('model.h5')

## load the encoders and scalar
with open('onehot_encoder_geo.pkl', 'rb') as file:
    onehot_encoder_geo = pickle.load(file)

with open('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)

with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender = pickle.load(file)


```

```python
input_data = {
    'CreditScore': 600,
    'Geography': 'France',  
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
geo_encoded = onehot_encoder_geo.transform([[input_data['Geography']]]).toarray()
one_hot_encoded_geo_df = pd.DataFrame(geo_encoded, columns=onehot_encoder_geo.get_feature_names_out(['Geography']))
one_hot_encoded_geo_df
```

```python
input_df = pd.DataFrame([input_data])
input_df
```

```python
input_df['Gender'] = label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
input_df = pd.concat([input_df.drop('Geography', axis=1), one_hot_encoded_geo_df], axis=1)
input_df
```

```python
#scaling the input data
input_scaled = scaler.transform(input_df)
input_scaled
```

```python
## Predict churn
prediction = model.predict(input_scaled)
prediction
```

```python
prediction_proba = prediction[0][0]
prediction_proba
```

```python
if prediction_proba > 0.5:
    print('The customer is likey to churn')
else:
    print('The customer is not likely to churn')
```
